```python
import sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define the correlation method. Defaulting to 'pearson' as it's the default for .corr()
method = 'pearson'


# Normalise names: "Current_ Stock" / "Current Stock" -> "current_stock"
df_clean.columns = (df_clean.columns.str.strip().str.lower()
              .str.replace(r"[\s_]+", "_", regex=True))

variables = [
    "current_physical_capital",
    "current_stock",
    "current_stock_finished",
    "revenue",
    "expenses",
    "profits",
    "loan_amount",
]
missing = [v for v in variables if v not in df_clean.columns]
if missing:
    raise SystemExit(f"Columns not found in data: {missing}\nAvailable: {list(df_clean.columns)}")

data = df_clean[variables].apply(pd.to_numeric, errors="coerce") # Use df_clean instead of df
corr = data.corr(method=method);

pretty = lambda s: s.replace("_", " ").title();

# 1) Revenue vs. everything else (single-column heatmap)
rev = corr[["revenue"]].drop(index="revenue").sort_values("revenue", ascending=False)
rev.index = [pretty(i) for i in rev.index]

fig, ax = plt.subplots(figsize=(5, 5))
sns.heatmap(rev, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1,
            linewidths=0.5, cbar_kws={"label": f"{method.title()} r"}, ax=ax)
ax.set_title("Correlation of Revenue with Other Variables")
ax.set_xlabel("")
ax.set_xticklabels(["Revenue"])
plt.tight_layout()
plt.savefig("revenue_correlation_heatmap.png", dpi=200)

# 2) Full matrix for context (optional)
full = corr.copy()
full.index = full.columns = [pretty(c) for c in full.columns]
fig, ax = plt.subplots(figsize=(8, 6.5))
sns.heatmap(full, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1,
            linewidths=0.5, square=True, ax=ax)
ax.set_title("Correlation Matrix")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("full_correlation_matrix.png", dpi=200)

print(rev.round(3))
plt.show()
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score, accuracy_score, confusion_matrix


# Ensembe Models
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df_clean.drop(columns=["loan_paid", "loan_fully_paid", "current_physical_capital"]) # features
y = df_clean["loan_paid"] # target
```

```python

X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
from imblearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python

model_dict = {
    "Logistic Regression with PowerTransformer": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000, class_weight="balanced"))
    ])
}
```

```python
name, lr_pipeline = list(model_dict.items())[0]

lr_pipeline.fit(X_train, y_train)
pred = lr_pipeline.predict(X_test)
prob = lr_pipeline.predict_proba(X_test)[:, 1]
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_auc_score, average_precision_score
```

```python

cv_scores = cross_val_score(
    lr_pipeline, X_train, y_train, cv=5, scoring="roc_auc"
)
```

```python
print("==Logistic Regression Report==")
print("ROC-AUC:", roc_auc_score(y_test, prob))
print("accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred))
print(f"Cross-validation ROC-AUC scores: {cv_scores}")
print(f"Mean ROC-AUC: {cv_scores.mean():.3f}")
print(f"Standard Deviation of ROC-AUC: {cv_scores.std():.3f}")
print(confusion_matrix(y_test, pred))
print("PR-AUC:", average_precision_score(y_test, prob))
```

```python
# SHAP TreeExplainer
explainer = shap.TreeExplainer(model_rf)
shap_values = explainer.shap_values(
    X_test,
    check_additivity=False
)

# For binary classification, class 1 = PAID
if isinstance(shap_values, list):
    shap_paid = shap_values[1]
else:
    shap_paid = shap_values[:, :, 1]

# Global importance
importance = np.abs(shap_paid).mean(axis=0)

shap_importance = pd.DataFrame({
    "Feature": X.columns,
    "Mean_Absolute_SHAP": importance
}).sort_values(
    "Mean_Absolute_SHAP",
    ascending=False
)

print(shap_importance)
```
